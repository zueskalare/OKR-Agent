# Matplotlib Styling Template Examples

This notebook demonstrates the usage of the highly customizable matplotlib plotting template based on the Plotly layout configuration.

## Features:
- Publication-ready plots with consistent styling
- **Automatic font scaling** based on figure size
- Custom color palettes
- Configurable fonts, sizes, and spacing
- Easy-to-use API for quick plots
- Advanced customization options

## Font Scaling:
- **Auto-scaling**: Fonts automatically adjust to figure size (larger figures get proportionally smaller fonts)
- **Manual scaling**: Use `styler.set_font_scale(0.5)` for half size, `styler.set_font_scale(2.0)` for double size
- **Disable auto-scaling**: Pass `auto_scale=False` to `create_publication_plot()`

## 1. Import Libraries and Setup

```python
import numpy as np
import matplotlib.pyplot as plt
from beaut4mpl.layout import (
    styler, 
    create_publication_plot, 
    apply_publication_style,
    color
)

# Apply global style settings
apply_publication_style()

print("✓ Matplotlib template loaded successfully!")
print(f"✓ Available color palettes: {list(color.keys())}")
```

## 1.5 Font Scaling Demonstration

The template now automatically scales fonts based on figure size, maintaining better proportions.

```python
# Compare different font scaling approaches
fig, axes = plt.subplots(1, 3, figsize=(24, 6))

x = np.linspace(0, 10, 50)
y = np.sin(x)

# Plot 1: Small figure with auto-scaling
styler.set_font_scale(1.0)
styler.auto_scale_fonts((8, 6))
ax1 = axes[0]
styler.style_axes(ax1)
ax1.plot(x, y, linewidth=3, color=styler.get_color_palette('nature', 1)[0])
styler.set_labels(ax1, 'X', 'Y', 'Small Fig (8×6)\nAuto Font')

# Plot 2: Medium figure with auto-scaling (current size)
styler.auto_scale_fonts((8, 6))  # This subplot is part of a 24×6 figure
ax2 = axes[1]
styler.style_axes(ax2)
ax2.plot(x, y, linewidth=3, color=styler.get_color_palette('nature', 2)[1])
styler.set_labels(ax2, 'X', 'Y', 'Medium Subplot\nAuto Font')

# Plot 3: Manual font scaling
styler.set_font_scale(0.7)  # Manually set smaller fonts
ax3 = axes[2]
styler.style_axes(ax3)
ax3.plot(x, y, linewidth=3, color=styler.get_color_palette('nature', 3)[2])
styler.set_labels(ax3, 'X', 'Y', 'Manual Scale\n(70% size)')

plt.tight_layout()
plt.show()

# Reset to default
styler.set_font_scale(1.0)
print("✓ Font scaling demonstrated. Notice how fonts stay proportional to figure size!")
```

## 2. Basic Line Plot with Custom Colors

A simple line plot demonstrating multiple series with the Spring_Pastels color palette.

```python
# Create figure using the template
fig, ax = create_publication_plot(
    xlabel="Compression Ratio",
    ylabel="Poisson's Ratio",
    title="Material Properties Comparison"
)

# Generate data
x = np.linspace(0, 2, 100)
colors_palette = styler.get_color_palette('Spring_Pastels', 4)

# Plot multiple lines
for i, (amplitude, phase, label) in enumerate([
    (0.3, 0, 'Material A'),
    (0.25, np.pi/4, 'Material B'),
    (0.2, np.pi/2, 'Material C'),
    (0.15, 3*np.pi/4, 'Material D')
]):
    y = amplitude * np.sin(2*np.pi*x + phase) + 0.35
    styler.plot_line(ax, x, y, label=label, color=colors_palette[i])

# Add legend
styler.add_legend(ax, ncol=2, loc='upper left', bbox_to_anchor=(0.01, 1.15))

plt.tight_layout()
plt.show()
```

## 3. Scatter Plot with Colormap

Scatter plot demonstrating the custom jianbian colormap with a publication-ready colorbar.

```python
# Create figure
fig, ax = create_publication_plot(
    xlabel="X Position (mm)",
    ylabel="Y Position (mm)",
    title="Stress Distribution Analysis"
)

# Generate random data simulating stress distribution
np.random.seed(42)
n_points = 200
x = np.random.randn(n_points) * 2
y = np.random.randn(n_points) * 2
stress = np.sqrt(x**2 + y**2) + np.random.randn(n_points) * 0.3

# Create scatter plot with custom colormap
scatter = styler.scatter_plot(
    ax, x, y, 
    c=stress, 
    cmap='custom_jianbian',
    s=600,  # larger points
    alpha=0.8,
    edgecolors='white',
    linewidths=1.5
)

# Add colorbar
styler.add_colorbar(
    scatter, 
    ax=ax, 
    label='Stress (MPa)',
    nticks=6,
    orientation='vertical'
)

plt.tight_layout()
plt.show()
```

## 4. Line Plot with Markers

Demonstrating line plots with different marker styles and the nature color palette.

```python
# Create figure
fig, ax = create_publication_plot(
    xlabel="Time (s)",
    ylabel="Displacement (mm)",
    title="Dynamic Loading Response"
)

# Generate data
x = np.linspace(0, 10, 20)
colors_nature = styler.get_color_palette('nature', 3)
markers = styler.get_markers(3)

# Plot with different markers
for i, (factor, label, marker, col) in enumerate(zip(
    [1.0, 1.5, 2.0],
    ['Load Case 1', 'Load Case 2', 'Load Case 3'],
    markers,
    colors_nature
)):
    y = factor * np.exp(-x/5) * np.cos(x)
    styler.plot_line(
        ax, x, y, 
        label=label, 
        color=col,
        marker=marker,
        markersize=15,
        linewidth=4,
        markeredgewidth=2,
        markeredgecolor='white'
    )

# Add legend
styler.add_legend(ax, ncol=3, loc='upper right')

plt.tight_layout()
plt.show()
```

## 5. Multiple Subplots

Creating a 2×2 grid of subplots with consistent styling.

```python
# Create 2x2 subplots
fig, axes = styler.create_figure(nrows=2, ncols=2, figsize=(20, 16))

# Flatten axes for easy iteration
axes = axes.flatten()

# Different plot types for each subplot
plot_configs = [
    {'title': 'Sine Wave', 'func': lambda x: np.sin(x), 'palette': 'monet'},
    {'title': 'Cosine Wave', 'func': lambda x: np.cos(x), 'palette': 'molandi'},
    {'title': 'Exponential Decay', 'func': lambda x: np.exp(-x/5), 'palette': 'Pink_Foam'},
    {'title': 'Gaussian', 'func': lambda x: np.exp(-x**2/10), 'palette': 'Grey_to_Red'}
]

x = np.linspace(-5, 5, 100)

for idx, (ax, config) in enumerate(zip(axes, plot_configs)):
    # Style the axes
    styler.style_axes(ax)
    
    # Set labels
    styler.set_labels(
        ax,
        xlabel='X Variable',
        ylabel='Y Response',
        title=config['title']
    )
    
    # Get colors
    colors_sub = styler.get_color_palette(config['palette'], 3)
    
    # Plot multiple series
    for i, (scale, col) in enumerate(zip([0.8, 1.0, 1.2], colors_sub)):
        y = scale * config['func'](x)
        styler.plot_line(
            ax, x, y,
            label=f'Scale {scale}',
            color=col,
            linewidth=3
        )
    
    # Add legend
    styler.add_legend(ax, fontsize=24, ncol=3, loc='upper right')

plt.tight_layout()
plt.show()
```

## 6. Advanced Customization

Demonstrating advanced customization with custom configuration dictionaries.

```python
# Create custom configuration
custom_config = {
    'axes_label': {
        'fontsize': 36,
        'color': '#2E4057',
    },
    'title': {
        'fontsize': 44,
        'color': '#2E4057',
    },
    'ticks': {
        'labelsize': 32,
        'color': '#2E4057',
    },
    'spines': {
        'linewidth': 3,
        'color': '#2E4057',
    },
    'grid': {
        'visible': True,
        'color': '#E8E8E8',
        'linestyle': '--',
        'alpha': 0.5,
    }
}

# Create figure
fig, ax = styler.create_figure(figsize=(14, 10))

# Apply custom styling to this specific axes
styler.style_axes(ax, config=custom_config)

# Generate data
theta = np.linspace(0, 4*np.pi, 200)
r = theta
x = r * np.cos(theta)
y = r * np.sin(theta)

# Create spiral with colormap
scatter = ax.scatter(
    x, y, 
    c=theta, 
    cmap='custom_jianbian',
    s=50,
    alpha=0.8,
    edgecolors='white',
    linewidths=0.5
)

# Set labels with custom config
styler.set_labels(
    ax,
    xlabel='X Coordinate',
    ylabel='Y Coordinate',
    title='Spiral Pattern with Custom Styling',
    config=custom_config
)

# Add colorbar
styler.add_colorbar(
    scatter,
    ax=ax,
    label='Angle (rad)',
    labelsize=36,
    ticklabelsize=32
)

plt.tight_layout()
plt.show()
```

## 7. Contour Plot with Colorbar

Creating a contour plot demonstrating heatmap visualization.

```python
# Create figure
fig, ax = create_publication_plot(
    xlabel="X Distance (mm)",
    ylabel="Y Distance (mm)",
    title="Temperature Distribution Field"
)

# Generate 2D data
x = np.linspace(-3, 3, 100)
y = np.linspace(-3, 3, 100)
X, Y = np.meshgrid(x, y)
Z = np.sin(np.sqrt(X**2 + Y**2)) * np.exp(-0.1 * (X**2 + Y**2))

# Create filled contour plot
contour = ax.contourf(X, Y, Z, levels=20, cmap='custom_jianbian', alpha=0.9)

# Add contour lines
contour_lines = ax.contour(X, Y, Z, levels=10, colors='white', 
                           linewidths=1, alpha=0.4)

# Add colorbar
cbar = styler.add_colorbar(
    contour,
    ax=ax,
    label='Temperature (°C)',
    nticks=8
)

plt.tight_layout()
plt.show()
```

## 8. Bar Plot Comparison

Creating a grouped bar chart with custom colors.

```python
# Create figure
fig, ax = create_publication_plot(
    xlabel="Material Type",
    ylabel="Strength (MPa)",
    title="Material Strength Comparison"
)

# Data
materials = ['Steel', 'Aluminum', 'Titanium', 'Carbon Fiber']
tensile = [400, 280, 900, 600]
compressive = [450, 300, 970, 500]
shear = [250, 180, 550, 350]

# Bar positions
x_pos = np.arange(len(materials))
width = 0.25

# Get colors
colors_bar = styler.get_color_palette('nature', 3)

# Create grouped bars
bars1 = ax.bar(x_pos - width, tensile, width, label='Tensile', 
               color=colors_bar[0], edgecolor='white', linewidth=2)
bars2 = ax.bar(x_pos, compressive, width, label='Compressive',
               color=colors_bar[1], edgecolor='white', linewidth=2)
bars3 = ax.bar(x_pos + width, shear, width, label='Shear',
               color=colors_bar[2], edgecolor='white', linewidth=2)

# Customize
ax.set_xticks(x_pos)
ax.set_xticklabels(materials, fontsize=32, fontweight='bold')
ax.set_ylim(0, 1100)

# Add legend
styler.add_legend(ax, ncol=3, loc='upper left')


plt.tight_layout()
plt.show()
```

## 9. Error Bars and Confidence Intervals

Demonstrating plots with error bars for uncertainty visualization.

```python
# Create figure
fig, ax = create_publication_plot(
    xlabel="Strain (%)",
    ylabel="Stress (MPa)",
    title="Stress-Strain Curves with Uncertainties"
)

# Generate data with noise
x = np.linspace(0, 5, 30)
colors_err = styler.get_color_palette('monet', 2)

for i, (E, label, col) in enumerate(zip([50, 80], ['Specimen A', 'Specimen B'], colors_err)):
    # Linear elastic response with some nonlinearity
    y = E * x * (1 - 0.02 * x)
    # Add uncertainty
    yerr = 0.05 * y + 2
    
    # Plot with error bars
    ax.errorbar(
        x, y,
        yerr=yerr,
        label=label,
        color=col,
        linewidth=4,
        marker='o',
        markersize=12,
        markeredgewidth=2,
        markeredgecolor='white',
        capsize=8,
        capthick=2,
        elinewidth=2,
        alpha=0.9
    )

# Add legend
styler.add_legend(ax, loc='upper left', fontsize=24)
styler.set_font_scale(4.0)

plt.tight_layout()
plt.show()
```

## 10. Polar Plot

Creating a polar plot with the template styling.

```python
# Create polar subplot
fig = plt.figure(figsize=(12, 12), dpi=100)
ax = fig.add_subplot(111, projection='polar')

# Generate data - directional strength measurements
angles = np.linspace(0, 2*np.pi, 8, endpoint=False)
strengths = [100, 85, 95, 110, 90, 105, 88, 98]

# Close the plot
angles = np.concatenate([angles, [angles[0]]])
strengths = np.concatenate([strengths, [strengths[0]]])

# Get colors
colors_polar = styler.get_color_palette('Pink_Foam', 3)

# Plot multiple datasets
for i, scale in enumerate([0.8, 1.0, 1.2]):
    scaled_strengths = np.array(strengths) * scale
    ax.plot(
        angles, scaled_strengths,
        linewidth=4,
        label=f'Condition {i+1}',
        color=colors_polar[i],
        marker='o',
        markersize=12,
        markeredgewidth=2,
        markeredgecolor='white'
    )
    ax.fill(angles, scaled_strengths, alpha=0.15, color=colors_polar[i])

# Customize polar plot
ax.set_theta_zero_location('N')
ax.set_theta_direction(-1)
ax.set_ylim(0, 140)
ax.set_title('Directional Strength Analysis', 
             fontsize=40, fontweight='bold', pad=30)
ax.tick_params(labelsize=28)
ax.grid(True, color='#E8E8E8', linewidth=1.5, alpha=0.7)

# Add legend
legend = ax.legend(
    fontsize=24,
    loc='upper right',
    bbox_to_anchor=(1.3, 1.1),
    frameon=True,
    facecolor='white',
    edgecolor='#666666',
    framealpha=1.0
)
for text in legend.get_texts():
    text.set_weight('bold')

plt.tight_layout()
plt.show()
```

## 12. Box Plot with Textures/Hatching

Demonstrate box plots with different hatch patterns for visual distinction.

```python
# Create figure
fig, ax = create_publication_plot(
    xlabel="Material Type",
    ylabel="Strength (MPa)",
    title="Material Strength Distribution with Hatch Patterns",
    figsize=(10, 6)
)

# Generate sample data for different materials
np.random.seed(42)
data_steel = np.random.normal(200, 20, 100)
data_aluminum = np.random.normal(150, 15, 100)
data_titanium = np.random.normal(250, 25, 100)
data_composite = np.random.normal(180, 18, 100)

data = [data_steel, data_aluminum, data_titanium, data_composite]
labels = ['Steel', 'Aluminum', 'Titanium', 'Composite']

# Get colors from palette
colors_box = styler.get_color_palette('Spring_Pastels', 4)

# Different hatch patterns for each box
hatches = ['///', '\\\\\\', '|||', '---']

# Create box plot with textures
bp = styler.box_plot(
    ax, 
    data, 
    labels=labels,
    colors=colors_box,
    hatches=hatches,
    showfliers=True
)

# Customize y-axis
ax.set_ylim(100, 320)

plt.tight_layout()
plt.show()

print("Box plot hatch patterns:")
print("Steel: ///  (forward diagonal)")
print("Aluminum: \\\\\\ (backward diagonal)")
print("Titanium: ||| (vertical lines)")
print("Composite: --- (horizontal lines)")
```

## 13. Bar Plot with Textures/Hatching

Demonstrate bar plots with hatch patterns for better distinction in black & white printing.

```python
# Create figure with subplots for comparison
fig, (ax1, ax2) = styler.create_subplots(1, 2, figsize=(14, 6))

# Sample data
materials = ['Steel', 'Al', 'Ti', 'Comp']
strengths = [200, 150, 250, 180]
x_pos = np.arange(len(materials))

# Get colors
colors_bar = styler.get_color_palette('Pink_Foam', 4)

# Left plot: Bar plot with colors and hatches
hatches_patterns = ['///', '\\\\\\', '|||', 'xxx']
for i, (pos, strength, color, hatch) in enumerate(zip(x_pos, strengths, colors_bar, hatches_patterns)):
    styler.bar_plot(
        ax1, 
        [pos], 
        [strength], 
        color=color, 
        hatch=hatch,
        label=materials[i]
    )

styler.set_labels(
    ax1,
    xlabel="Material Type",
    ylabel="Strength (MPa)",
    title="Bar Plot with Different Hatch Patterns"
)
ax1.set_xticks(x_pos)
ax1.set_xticklabels(materials)
ax1.set_ylim(0, 280)
styler.add_legend(ax1, loc='upper right')

# Right plot: Grouped bars with hatches
width = 0.35
conditions = ['Room Temp', 'High Temp']
strengths_rt = [200, 150, 250, 180]
strengths_ht = [180, 135, 225, 162]

bars1 = styler.bar_plot(
    ax2,
    x_pos - width/2,
    strengths_rt,
    color=colors_bar[0],
    hatch='///',
    label='Room Temp',
    width=width
)

bars2 = styler.bar_plot(
    ax2,
    x_pos + width/2,
    strengths_ht,
    color=colors_bar[2],
    hatch='\\\\\\',
    label='High Temp',
    width=width
)

styler.set_labels(
    ax2,
    xlabel="Material Type",
    ylabel="Strength (MPa)",
    title="Grouped Bar Plot with Hatches"
)
ax2.set_xticks(x_pos)
ax2.set_xticklabels(materials)
ax2.set_ylim(0, 280)
styler.add_legend(ax2, loc='upper right')

plt.tight_layout()
plt.show()

print("\nAvailable hatch patterns:")
print("'/' - forward diagonal lines")
print("'\\' - backward diagonal lines (use '\\\\\\' in code)")
print("'|' - vertical lines")
print("'-' - horizontal lines")
print("'+' - crossed lines")
print("'x' - diagonal crossed lines")
print("'o' - small circles")
print("'O' - large circles")
print("'.' - dots")
print("'*' - stars")
print("\nCombine patterns for denser textures: '//', '|||', 'xxx', etc.")
```
